```python
def while_multiply(a,b):
    result = 0
    while(b > 0):
        
        result += a 
        b -= 1
    return result
print(while_multiply(6,7))
```

# a and b are the parameters. b starts off as 7. the line "b -= 1" subtracts 1 each time the loop runs in order to keep track of how many times 6 is added to itself, until it becomes 0 and gives the product.

```python
def for_multiply (a,b):
    result = 0
    for i in range(b):
        result = result + a
    return result
print(for_multiply(6,7))
```
