# I used the same cutoffs as the book. below 15 is cold, 15-25 (inclusive) is warm, and greater than 25 is hot. 

```python
temp = [10, 16, 23, 28, 14, 22, 30]
for i in temp:
        if i > 25:
            print([i],"is hot")
        elif i >= 15 and i <= 25:
            print([i],"is warm")
        else:
            print ([i], "is cold")
    
```
