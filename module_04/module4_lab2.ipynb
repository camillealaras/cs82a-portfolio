```python
import pandas as pd
df = pd.read_csv("allbacks.csv")

df["cover"] = pd.Categorical(
    df["cover"],
    categories=["hb", "pb"],
    ordered=True
)
print(df["cover"].cat.codes.median())
```

# The book type that falls in the middle when arranged in order by cover type is the paperback type.

```python

print(df["weight"].mean())
```

# The average weight of both hardcovers and paperbacks is ~706.6g.

```python
print(df.groupby("cover")["weight"].mean())
```

# The mean weight of hardcovers is considerably larger than the mean weight of the paperback books, which suggests that the average hardcover book weighs more than the average paperback. However, the mean is not reisistant to outliers so this might not necessarily be correct.
