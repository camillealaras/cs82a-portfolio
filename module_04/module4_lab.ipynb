```python
import pandas as pd
df = pd.read_csv("sales_clean.csv")
df.shape
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())
```

```python
print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

# I would use the median to report for price, since the mean of price is considerably larger than the median. The difference could be because of outliers that skew the data to the left, which leaves the median as more reliable because it's more resistant outliers.

```python
df["price"].plot(kind="hist", bins=20)
```

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
df.plot.scatter(x="price", y="revenue")
df.plot.scatter(x="qty", y="revenue")
```

# The relationship between price and revenue has a strong correlation, with coefficient being 0.817856. However, when looking at the scatterplot, it appears that the strength of the correlation could be falsely inflated by outliers. Likewise, there is a low correlation (0.330733) between quantity and revenue, however, the scatterplot also seems to show that this correlation could be attributed to outliers rather than an actual correlation.
