```python
import pandas as pd
df = pd.read_csv("usarrests.csv")


fill_value = df["Assault"].median()
df["Assault"] = df["Assault"].fillna(fill_value)

df.isna().sum()
```

# To fix Georgia's missing assault value, I replaced it with the median. I used the median because you're supposed to use the median to replace a missing value when the data contains outliers and is not symmetrical. South Carolina's value was an outlier, so I had to use the median and not the mean.
